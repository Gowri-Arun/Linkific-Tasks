# Data Visualization using Matplotlib and Seaborn

This notebook visualizes the student dataset from Day 5.

## Load the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

students = pd.read_csv('../Day-5/students.csv')
students.head()
```

## Dataset Information

```python
students.info()
```

## Bar Chart: Number of Students by Age

This chart compares the number of students in each age group. Age has only four categories, so each category remains readable.

```python
age_counts = students['Age'].value_counts().sort_index()

plt.figure(figsize=(7, 4))
sns.barplot(x=age_counts.index, y=age_counts.values, hue=age_counts.index, palette='Blues_d', legend=False)
plt.title('Number of Students by Age')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.ylim(0, age_counts.max() + 1)
plt.tight_layout()
plt.show()
```

## Line Chart: Maths Scores in Ascending Order

This chart orders students by their Maths score to show the progression from the lowest to highest recorded score. It does not represent a time trend. The row with a missing Maths score is excluded.

```python
maths_scores = students[['Name', 'Maths']].dropna().sort_values('Maths')

plt.figure(figsize=(8, 4))
plt.plot(maths_scores['Name'], maths_scores['Maths'], marker='o', color='teal', label='Maths score')
plt.title('Maths Scores in Ascending Order')
plt.xlabel('Student')
plt.ylabel('Maths Score')
plt.ylim(0, 100)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## Histogram: Attendance Distribution

This histogram shows how the recorded attendance values are distributed. The missing attendance value is excluded.

```python
attendance = students['Attendance'].dropna()

plt.figure(figsize=(7, 4))
sns.histplot(attendance, bins=5, color='coral', edgecolor='black')
plt.title('Distribution of Attendance')
plt.xlabel('Attendance')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

## Pie Chart: Proportion of Students by Age

This chart shows each age group's share of all eight students.

```python
plt.figure(figsize=(6, 6))
plt.pie(age_counts.values, labels=age_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Proportion of Students by Age')
plt.legend(title='Age', loc='best')
plt.tight_layout()
plt.show()
```
