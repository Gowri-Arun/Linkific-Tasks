# NumPy Fundamentals

Beginner-level examples of common NumPy array operations.

```python
import numpy as np
```

```python
one_d = np.array([10, 20, 30, 40])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
print(one_d)
print(two_d)
```

```python
print('Dimensions:', two_d.ndim)
print('Shape:', two_d.shape)
print('Size:', two_d.size)
print('Data type:', two_d.dtype)
```

```python
print('First value:', one_d[0])
print('Second row, third value:', two_d[1, 2])
print('Slice:', one_d[1:3])
print('First column:', two_d[:, 0])
```

```python
print('Column sums:', two_d.sum(axis=0))
print('Row sums:', two_d.sum(axis=1))
print('Mean:', two_d.mean())
print('Maximum:', two_d.max())
print('Minimum:', two_d.min())
```

```python
numbers = np.array([1, 2, 3])
print('Add:', numbers + 5)
print('Multiply:', numbers * 2)
print('Square:', numbers ** 2)
print('Broadcasting:', two_d + np.array([10, 20, 30]))
```

```python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])
print('Python list * 2:', python_list * 2)
print('NumPy array * 2:', numpy_array * 2)
```
